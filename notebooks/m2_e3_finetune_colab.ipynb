## 5. Example Computation and Visualization
Toy forward pass + quick chart to confirm environment works. Replace with full training loops by calling scripts in `/content/drive/MyDrive/.../scripts/m2_e3`.

```python
import torch
import matplotlib.pyplot as plt

# Tiny linear regression demo
x = torch.linspace(0, 1, steps=50).unsqueeze(1)
y = 3 * x + 0.2 + 0.05 * torch.randn_like(x)
model = torch.nn.Linear(1, 1)
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
loss_fn = torch.nn.MSELoss()

for _ in range(200):
    optimizer.zero_grad()
    pred = model(x)
    loss = loss_fn(pred, y)
    loss.backward()
    optimizer.step()

print('Final loss:', loss.item())

with torch.no_grad():
    y_hat = model(x)

plt.figure(figsize=(5, 3))
plt.scatter(x.numpy(), y.numpy(), label='data', s=10)
plt.plot(x.numpy(), y_hat.numpy(), color='red', label='fit')
plt.legend()
plt.title('Sanity check regression')
plt.show()
```

## 6. Prepare data splits
Runs data prep to create train/val/test splits and labels.json. Adjust ratios if needed.

```python
import json, subprocess, sys
from pathlib import Path

SPLITS_DIR = DATA_DIR / 'splits'
SPLITS_DIR.mkdir(parents=True, exist_ok=True)

cmd = [
    sys.executable,
    'scripts/m2_e3/data_prep.py',
    '--data', str(GOLD_PATH),
    '--output-dir', str(SPLITS_DIR),
    '--train-ratio', '0.8',
    '--val-ratio', '0.1',
    '--seed', '13',
]
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)

stats_path = SPLITS_DIR / 'stats.json'
if stats_path.exists():
    print('Stats:')
    print(stats_path.read_text())
else:
    print('stats.json not found; check output above.')
```

## 7. Train intent classifier (MiniLM)
Uses scripts/m2_e3/train_intent.py with the splits. Adjust epochs/batch to fit GPU memory.

```python
INTENT_DIR = ARTIFACTS_DIR / 'intent'
INTENT_DIR.mkdir(parents=True, exist_ok=True)

MODEL_NAME = os.environ.get('INTENT_MODEL_NAME', 'sentence-transformers/all-MiniLM-L6-v2')

cmd = [
    sys.executable,
    'scripts/m2_e3/train_intent.py',
    '--splits-dir', str(SPLITS_DIR),
    '--output-dir', str(INTENT_DIR),
    '--model-name', MODEL_NAME,
    '--num-epochs', '3',
    '--batch-size', '16',
    '--lr', '5e-5',
    '--logging-steps', '100',
]
print('Running:', ' '.join(cmd))
proc = subprocess.run(cmd, capture_output=True, text=True)
print('stdout:\n', proc.stdout)
print('stderr:\n', proc.stderr)
if proc.returncode != 0:
    raise RuntimeError(f"train_intent failed with code {proc.returncode}")

print('Intent artifacts:', list(INTENT_DIR.iterdir()))
```

## 8. Train parser (Flan-T5-small)
Uses scripts/m2_e3/train_parser_t5.py on the same splits. Lower batch/max target length if you hit OOM.

```python
PARSER_DIR = ARTIFACTS_DIR / 'parser'
PARSER_DIR.mkdir(parents=True, exist_ok=True)

cmd = [
    sys.executable,
    'scripts/m2_e3/train_parser_t5.py',
    '--splits-dir', str(SPLITS_DIR),
    '--output-dir', str(PARSER_DIR),
    '--num-epochs', '3',
    '--batch-size', '4',
    '--lr', '3e-5',
    '--max-target-length', '256',
    '--logging-steps', '100',
]
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)

print('Parser artifacts:', list(PARSER_DIR.iterdir()))
```

## 9. Inference on test split (model + rules)
Loads trained bundles, runs a few test rows through the model parser+intent (with rule fallback on validation errors), and also shows the pure rules output for comparison.

```python
from experiments.m2_e3_parse import run_parse
import json
from pathlib import Path

test_path = SPLITS_DIR / 'test.jsonl'
if not test_path.exists():
    raise FileNotFoundError(f"Missing test split at {test_path}")

with test_path.open() as f:
    test_rows = [json.loads(line) for line in f][:5]
print('Loaded test rows:', len(test_rows))

bundles = {
    'intent': run_parse.load_intent_bundle(INTENT_DIR),
    'parser': run_parse.load_parser_bundle(PARSER_DIR),
}

for row in test_rows:
    model_pred, justification = run_parse.parse_row(row, bundles, threshold=0.5, fallback_on_error=True)
    rules_pred = run_parse.parse_row_rules(row)
    print('\n====', row['id'], '====')
    print('MODEL PARSER OUTPUT:')
    print(json.dumps(model_pred, indent=2))
    print('Validation:', justification.get('validation'))
    print('RULES OUTPUT:')
    print(json.dumps(rules_pred, indent=2))
```
